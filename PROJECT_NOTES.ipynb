## 20. Common Errors and Solutions

### Error: `git is not recognized`

**Cause:** Git is not installed or VS Code was not restarted after installation.

**Solution:** Install Git for Windows, restart VS Code, and run:

```powershell
git --version
```

### Error: `Author identity unknown`

**Cause:** Git username and email are not configured.

**Solution:**

```powershell
git config --global user.name "Sarath Kumar"
git config --global user.email "YOUR-GITHUB-EMAIL"
```

### Error: `remote origin already exists`

**Cause:** A GitHub remote is already connected.

**Check the existing remote:**

```powershell
git remote -v
```

**Change an incorrect remote:**

```powershell
git remote set-url origin https://github.com/YOUR-USERNAME/whatsapp-bot-playwright.git
```

### Error: Rejected push because the remote contains work

**Cause:** The GitHub repository was created with a README or another file.

**Safe solution:**

```powershell
git pull --rebase origin main
git push -u origin main
```

Resolve any reported conflict before pushing again.

### Private file was staged accidentally

Remove it from the staging area without deleting the local file:

```powershell
git restore --staged contacts.xlsx
```

Then correct `.gitignore` and run:

```powershell
git status
```

### Private file was already committed

Do not push the commit. Remove it from Git tracking:

```powershell
git rm --cached contacts.xlsx
```

Add it to `.gitignore`, commit the correction, and verify the history before publishing.

If a password, token, QR code, or active session was committed, treat it as exposed and replace or revoke it.
